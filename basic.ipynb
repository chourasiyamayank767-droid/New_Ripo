```python
arr1=np.linspace(12,22,num=10, dtype=int)
```

```python
# for population variance
def variance_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
        calculate_variance=total/(len(args))
        return calculate_variance,calculate_variance**(1/2)
    
```

```python
variance_std(*arr1)
```

```python
# for sample variance
def variance_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
        calculate_variance=total/(len(args)-1)
        return calculate_variance,calculate_variance**(1/2)
```

```python
variance_std(*arr1)
```

```python
hours=[1,2,3,4,5,6]
marks=[30,40,50,55,65,75]

```

```python
plt.figure(figsize=(5,4))
plt.scatter(hours,marks,color='red',linewidth=1,marker='+')
plt.xlabel('no of hours')
plt.ylabel('hours vs marks')
plt.show()
```
